# 💎 Diamond Price & Cluster Prediction — Debug Notebook

This notebook mirrors the Streamlit app's logic exactly, so you can debug the
feature engineering, encoding, and preprocessing pipeline step by step.

**Pipeline recap:**
- Price model (XGBoost): predicts directly on **raw** engineered features (no scaling)
- Cluster model (KMeans): PowerTransformer → StandardScaler → PCA → KMeans

Place this notebook in the same folder as the `models/` directory (or update `MODELS_DIR` below).

````
import pickle
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 150)

MODELS_DIR = Path("models")  # update this if your models folder is elsewhere
assert MODELS_DIR.exists(), f"Models folder not found at {MODELS_DIR.resolve()}"
````

## 1. Load all trained artifacts

````
with open(MODELS_DIR / "xgb_model.pkl", "rb") as f:
    xgb_model = pickle.load(f)

with open(MODELS_DIR / "feature_columns.pkl", "rb") as f:
    FEATURE_COLUMNS = pickle.load(f)

with open(MODELS_DIR / "standard_scaler.pkl", "rb") as f:
    scaler = pickle.load(f)

with open(MODELS_DIR / "power_transformer.pkl", "rb") as f:
    power_transformer = pickle.load(f)

with open(MODELS_DIR / "pca.pkl", "rb") as f:
    pca = pickle.load(f)

with open(MODELS_DIR / "clustering_model.pkl", "rb") as f:
    kmeans = pickle.load(f)

with open(MODELS_DIR / "cluster_names.pkl", "rb") as f:
    cluster_names = pickle.load(f)

print("xgb_model:         ", type(xgb_model).__name__)
print("FEATURE_COLUMNS:   ", FEATURE_COLUMNS)
print("scaler features:   ", list(scaler.feature_names_in_))
print("power_transformer: ", list(power_transformer.feature_names_in_))
print("pca n_components_: ", pca.n_components_)
print("kmeans n_clusters: ", kmeans.n_clusters)
print("cluster_names:     ", cluster_names)
````

## 2. Encoding maps

These must match the `OrdinalEncoder` categories used during training **exactly**:

```python
OrdinalEncoder(categories=[
    ['Fair','Good','Very Good','Premium','Ideal'],
    ['J','I','H','G','F','E','D'],
    ['I1','SI2','SI1','VS2','VS1','VVS2','VVS1','IF']
])
```

````
CUT_MAP = {"Fair": 0, "Good": 1, "Very Good": 2, "Premium": 3, "Ideal": 4}
COLOR_MAP = {"J": 0, "I": 1, "H": 2, "G": 3, "F": 4, "E": 5, "D": 6}
CLARITY_MAP = {"I1": 0, "SI2": 1, "SI1": 2, "VS2": 3, "VS1": 4, "VVS2": 5, "VVS1": 6, "IF": 7}

POWER_COLS = list(power_transformer.feature_names_in_)
print("Columns that get power-transformed before scaling:", POWER_COLS)
````

## 3. Feature engineering function

Builds the 12-column feature row from the raw user inputs:

- `depth_pct = depth / ((length + width) / 2) * 100`
- `dimension_ratio = (length * width) / (depth * 2)`
- `volume = length * width * depth`
- `face_area = length * width`

````
def build_feature_row(carat, cut, color, clarity, length, width, depth, table, verbose=False):
    cut_enc = CUT_MAP[cut]
    color_enc = COLOR_MAP[color]
    clarity_enc = CLARITY_MAP[clarity]

    depth_pct = depth / ((length + width) / 2) * 100
    dimension_ratio = (length * width) / (depth * 2)
    volume = length * width * depth
    face_area = length * width

    row = {
        "carat": carat,
        "cut": cut_enc,
        "color": color_enc,
        "clarity": clarity_enc,
        "depth_pct": depth_pct,
        "table": table,
        "length": length,
        "width": width,
        "depth": depth,
        "dimension_ratio": dimension_ratio,
        "volume": volume,
        "face_area": face_area,
    }
    df = pd.DataFrame([row])[FEATURE_COLUMNS]

    if verbose:
        print("Engineered feature row:")
        print(df.T.rename(columns={0: "value"}))

    return df
````

## 4. Price prediction (XGBoost — raw features, no scaling)

````
def predict_price(raw_df, verbose=False):
    pred = xgb_model.predict(raw_df)[0]
    if verbose:
        print(f"Predicted price: Rs. {pred:,.2f}")
    return float(pred)
````

## 5. Cluster prediction (PowerTransformer → StandardScaler → PCA → KMeans)

Each intermediate step is printed so you can inspect where things might drift
if your predictions don't match the app.

````
def predict_cluster(raw_df, verbose=False):
    df = raw_df.copy()

    # Step 1: power-transform the skewed subset of columns
    df[POWER_COLS] = power_transformer.transform(df[POWER_COLS])
    if verbose:
        print("--- After PowerTransformer ---")
        print(df)

    # Step 2: standard-scale all 12 columns (same order the scaler was fit on)
    scaled = scaler.transform(df[FEATURE_COLUMNS])
    if verbose:
        print("--- After StandardScaler ---")
        print(pd.DataFrame(scaled, columns=FEATURE_COLUMNS))

    # Step 3: PCA
    pca_vec = pca.transform(scaled)
    if verbose:
        print("--- After PCA ---")
        print(pca_vec)

    # Step 4: KMeans
    cluster_id = int(kmeans.predict(pca_vec)[0])
    cluster_label = cluster_names.get(cluster_id, f"Cluster {cluster_id}").replace("_", " ")

    if verbose:
        print(f"\nPredicted cluster: {cluster_id} -> {cluster_label}")

    return cluster_id, cluster_label, pca_vec
````

## 6. Try it out

````
raw_df = build_feature_row(
    carat=0.70,
    cut="Ideal",
    color="E",
    clarity="VS1",
    length=5.70,
    width=5.72,
    depth=3.53,
    table=57.5,
    verbose=True,
)
````

````
price = predict_price(raw_df, verbose=True)
````

````
cluster_id, cluster_label, pca_vec = predict_cluster(raw_df, verbose=True)
````

## 7. Batch debugging — run several test diamonds at once

````
test_diamonds = [
    dict(carat=0.30, cut="Ideal",     color="D", clarity="IF",  length=4.30, width=4.32, depth=2.68, table=56.0),
    dict(carat=0.70, cut="Very Good", color="G", clarity="VS2", length=5.70, width=5.72, depth=3.53, table=58.0),
    dict(carat=1.50, cut="Premium",   color="H", clarity="SI1", length=7.30, width=7.28, depth=4.51, table=59.0),
    dict(carat=2.50, cut="Fair",      color="J", clarity="I1",  length=8.60, width=8.55, depth=5.40, table=62.0),
]

results = []
for params in test_diamonds:
    raw = build_feature_row(**params)
    p = predict_price(raw)
    cid, cname, _ = predict_cluster(raw)
    results.append({**params, "predicted_price_inr": p, "cluster_id": cid, "cluster_name": cname})

pd.DataFrame(results)
````

## 8. Sanity checks

Quick checks you can use while debugging — e.g. confirming column order matches
what each fitted object expects, and that no NaNs sneak into the pipeline.

````
raw_df = build_feature_row(0.70, "Ideal", "E", "VS1", 5.70, 5.72, 3.53, 57.5)

assert list(raw_df.columns) == FEATURE_COLUMNS, "Column order mismatch vs feature_columns.pkl!"
assert list(raw_df.columns) == list(scaler.feature_names_in_), "Column order mismatch vs scaler!"
assert raw_df.isna().sum().sum() == 0, "NaNs found in engineered features!"

print("All checks passed.")
````
